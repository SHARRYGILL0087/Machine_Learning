# Ordinal Encoding

1. SkLearn    2. Map Fuction -> Techniques

1. Sklearn

```python
odf = pd.DataFrame({"Size" : ["s","m","l","xl","s","m","l","xl","s","m","l","xl"]})
odf.head(4)
```

```python
ord_data = [["s","m","l","xl"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=ord_data)
oe.fit(odf[["Size"]])
```

```python
odf["en_Size"] = oe.transform(df[["Size"]])
odf
```

# Map Function

```python
ord_data1 = {"s" : 0 , "m" : 1,"l" : 2 , "xl" : 3}
```

```python
odf["Size_en_map"] = odf["Size"].map(ord_data1)
odf
```

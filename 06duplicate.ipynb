# HANDLING DUPLICATE DATA

```python
import pandas as pd
```

```python
data = {"name":["a","b","c","d","a","c"],"eng":[2,3,4,5,2,8],"hindi":[6,5,4,7,6,2]}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.duplicated()
```

```python
# df.drop_duplicates(keep="first")
df.drop_duplicates(inplace=True)
```

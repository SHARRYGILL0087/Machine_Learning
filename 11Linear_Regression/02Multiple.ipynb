```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python

dataset = pd.read_csv(r"C:\Users\SHAMSHER\Downloads\salaryPredict.csv")
dataset.head(3)
```

```python
dataset.shape
```

Check relation between age and Experience with Salary

```python
sns.pairplot(data=dataset)
plt.show()
```

```python
sns.heatmap(data=dataset.corr(),annot=True)
plt.show()
```

```python
x = dataset.iloc[:, [0, 2]]
y = dataset["Salary"]
```

```python
x.ndim
```

```python
dataset.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```

```python
# y = m1x1 + m2x2 + c
```
